```python
DATA_PATH = "/content/drive/MyDrive/EmployeeAttrition/Employee_Dataset.csv"
```

```python
import pandas as pd

data = pd.read_csv(DATA_PATH)
```

```python
import numpy as np
```

```python
age_array = np.array(data["Age"])

print(age_array)
```

```python
print("Shape:", age_array.shape)
print("Dimensions:", age_array.ndim)
print("Number of values:", age_array.size)
print("Data type:", age_array.dtype)
```

```python
print(age_array[0])
print(age_array[1])
print(age_array[2])
```

```python
print("Average age:", np.mean(age_array))
print("Maximum age:", np.max(age_array))
print("Minimum age:", np.min(age_array))
print("Median age:", np.median(age_array))
print("Standard deviation:", np.std(age_array))
```

```python
print(age_array[0:5])
```

```python
print(age_array[:10])
```

```python
age_plus_one = age_array + 1

print(age_plus_one[:10])
```

```python
older_employees = age_array[age_array > 40]

print(older_employees)
```

```python
print("First 10 ages:")
print(age_array[:10])

print("\nAverage age:")
print(np.mean(age_array))

print("\nEmployees older than 40:")
print(np.sum(age_array > 40))
```

```python
print("Total employees:", age_array.size)
print("Average age:", np.mean(age_array))
print("Youngest employee:", np.min(age_array))
print("Oldest employee:", np.max(age_array))
print("Median age:", np.median(age_array))
```

```python
below_30 = np.sum(age_array < 30)

print("Employees below 30:", below_30)
```

```python
between_30_40 = np.sum((age_array >= 30) & (age_array <= 40))

print("Employees between 30 and 40:", between_30_40)
```

```python
above_40 = np.sum(age_array > 40)

print("Employees above 40:", above_40)
```

```python
total_age = np.sum(age_array)

print("Total age:", total_age)
```

```python
age_std = np.std(age_array)

print("Age standard deviation:", age_std)
```
