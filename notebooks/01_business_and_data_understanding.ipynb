```python
ages = df["Age"].to_numpy()

print("Type:", type(ages))
print("Dimensions:", ages.ndim)
print("Shape:", ages.shape)
print("Size:", ages.size)
print("Average Age:", np.mean(ages))
print("Minimum Age:", np.min(ages))
print("Maximum Age:", np.max(ages))
print("Median Age:", np.median(ages))
```

```python
employees_above_40 = ages[ages > 40]

print("Employees above 40:", employees_above_40.size)
```

```python
employees_below_30 = ages[ages < 30]

print("Employees below 30:", employees_below_30.size)
```
