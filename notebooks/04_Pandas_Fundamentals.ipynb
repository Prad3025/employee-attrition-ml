```python
data.groupby("Department")["MonthlyIncome"].agg(
    ["mean", "min", "max"]
)
```

```python
data.groupby("Department").size()
```

```python
data.groupby("Department")["Attrition"].value_counts()
```

```python
data["Attrition_Flag"] = data["Attrition"].map({
    "Yes": 1,
    "No": 0
})
```

```python
data[["Attrition", "Attrition_Flag"]].head(10)
```

```python
department_attrition = (
    data.groupby("Department")["Attrition_Flag"]
    .mean()
    * 100
)

print(department_attrition)
```

```python
department_summary = data.groupby("Department").agg(
    Employees=("EmployeeNumber", "count"),
    AverageIncome=("MonthlyIncome", "mean"),
    AverageAge=("Age", "mean"),
    AttritionRate=("Attrition_Flag", "mean")
)

department_summary["AttritionRate"] = (
    department_summary["AttritionRate"] * 100
)

department_summary
```

```python
data["JobRole"].value_counts()
```

```python
jobrole_attrition = (
    data.groupby("JobRole")["Attrition_Flag"]
    .mean()
    * 100
)

print(jobrole_attrition)
```

```python
data["OverTime"].value_counts()
```

```python
overtime_attrition = (
    data.groupby("OverTime")["Attrition_Flag"]
    .mean()
    * 100
)

print(overtime_attrition)
```

```python
satisfaction_attrition = (
    data.groupby("JobSatisfaction")["Attrition_Flag"]
    .mean()
    * 100
)

print(satisfaction_attrition)
```

```python
worklife_attrition = (
    data.groupby("WorkLifeBalance")["Attrition_Flag"]
    .mean()
    * 100
)

print(worklife_attrition)
```
