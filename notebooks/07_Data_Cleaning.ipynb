```python
import pandas as pd
import numpy as np

data = pd.read_csv(
    "/content/drive/MyDrive/EmployeeAttrition/Employee_Dataset.csv"
)

print(data.shape)
```

```python
data.isnull().sum()
```

```python
print("Total missing values:", data.isnull().sum().sum())
```

```python
print("Duplicate rows:", data.duplicated().sum())
```

```python
constant_columns = []

for col in data.columns:
    unique_values = data[col].nunique()

    if unique_values == 1:
        constant_columns.append(col)

print("Constant columns:")
print(constant_columns)
```

```python
print("Number of rows:", len(data))
print("Unique EmployeeNumber:", data["EmployeeNumber"].nunique())
```

```python
categorical_columns = data.select_dtypes(
    include=["object"]
).columns

print(categorical_columns)
```

```python
numerical_columns = data.select_dtypes(
    include=["int64", "float64"]
).columns

print(numerical_columns)
```

```python
cleaned_data = data.copy()
```

```python
cleaned_data = cleaned_data.drop(
    columns=constant_columns
)
```

```python
cleaned_data = cleaned_data.drop(
    columns=["EmployeeNumber"]
)
```

```python
print("Original shape:", data.shape)
print("Cleaned shape:", cleaned_data.shape)
```

```python
import os

processed_path = (
    "/content/drive/MyDrive/"
    "EmployeeAttrition/processed"
)

os.makedirs(processed_path, exist_ok=True)
```

```python
cleaned_data.to_csv(
    processed_path + "/Employee_Dataset_Cleaned.csv",
    index=False
)

print("Cleaned dataset saved successfully!")
```
