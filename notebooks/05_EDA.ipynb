```python
import pandas as pd
import numpy as np

data = pd.read_csv(
    "/content/drive/MyDrive/EmployeeAttrition/Employee_Dataset.csv"
)

print(data.shape)
```

```python
data["Attrition"].value_counts()
```

```python
data["Attrition"].value_counts(normalize=True) * 100
```

```python
data["Attrition_Flag"] = data["Attrition"].map({
    "Yes": 1,
    "No": 0
})
```

```python
data[["Attrition", "Attrition_Flag"]].head()
```

```python
attrition_rate = data["Attrition_Flag"].mean() * 100

print("Overall Attrition Rate:", round(attrition_rate, 2), "%")
```

```python
overtime_analysis = (
    data.groupby("OverTime")["Attrition_Flag"]
    .mean()
    * 100
)

print(overtime_analysis.round(2))
```

```python
import matplotlib.pyplot as plt
```

```python
overtime_analysis.plot(kind="bar")

plt.title("Attrition Rate by Overtime")
plt.xlabel("Overtime")
plt.ylabel("Attrition Rate (%)")
plt.xticks(rotation=0)
plt.show()
```

```python
department_attrition = (
    data.groupby("Department")["Attrition_Flag"]
    .mean()
    * 100
)

print(department_attrition.round(2))
```

```python
department_attrition.plot(kind="bar")

plt.title("Attrition Rate by Department")
plt.xlabel("Department")
plt.ylabel("Attrition Rate (%)")
plt.xticks(rotation=45)
plt.show()
```

```python
jobrole_attrition = (
    data.groupby("JobRole")["Attrition_Flag"]
    .mean()
    * 100
)

print(jobrole_attrition.round(2))
```

```python
jobrole_attrition = jobrole_attrition.sort_values(ascending=False)

print(jobrole_attrition.round(2))
```

```python
jobrole_attrition.plot(kind="bar")

plt.title("Attrition Rate by Job Role")
plt.xlabel("Job Role")
plt.ylabel("Attrition Rate (%)")
plt.xticks(rotation=45, ha="right")
plt.show()
```

```python
numerical_summary = data.groupby("Attrition")[
    [
        "Age",
        "MonthlyIncome",
        "DistanceFromHome",
        "TotalWorkingYears",
        "YearsAtCompany",
        "YearsInCurrentRole",
        "YearsSinceLastPromotion",
        "YearsWithCurrManager"
    ]
].mean()

numerical_summary.round(2)
```
