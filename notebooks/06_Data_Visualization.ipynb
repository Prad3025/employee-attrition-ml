```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

data = pd.read_csv(
    "/content/drive/MyDrive/EmployeeAttrition/Employee_Dataset.csv"
)

data["Attrition_Flag"] = data["Attrition"].map({
    "Yes": 1,
    "No": 0
})

print(data.shape)
```

```python
attrition_count = data["Attrition"].value_counts()

attrition_count.plot(kind="bar")

plt.title("Employee Attrition")
plt.xlabel("Attrition")
plt.ylabel("Number of Employees")
plt.xticks(rotation=0)
plt.show()
```

```python
attrition_percentage = (
    data["Attrition"]
    .value_counts(normalize=True)
    * 100
)

print(attrition_percentage.round(2))
```

```python
plt.figure(figsize=(6, 6))

plt.pie(
    attrition_percentage,
    labels=attrition_percentage.index,
    autopct="%1.1f%%"
)

plt.title("Employee Attrition Percentage")
plt.show()
```

```python
department_attrition = (
    data.groupby("Department")["Attrition_Flag"]
    .mean()
    * 100
)

department_attrition = department_attrition.sort_values(
    ascending=False
)

print(department_attrition.round(2))
```

```python
plt.figure(figsize=(8, 5))

department_attrition.plot(kind="bar")

plt.title("Attrition Rate by Department")
plt.xlabel("Department")
plt.ylabel("Attrition Rate (%)")
plt.xticks(rotation=45)
plt.show()
```

```python
overtime_attrition = (
    data.groupby("OverTime")["Attrition_Flag"]
    .mean()
    * 100
)

print(overtime_attrition.round(2))
```

```python
plt.figure(figsize=(7, 5))

overtime_attrition.plot(kind="bar")

plt.title("Attrition Rate by Overtime")
plt.xlabel("Overtime")
plt.ylabel("Attrition Rate (%)")
plt.xticks(rotation=0)
plt.show()
```

```python
jobrole_attrition = (
    data.groupby("JobRole")["Attrition_Flag"]
    .mean()
    * 100
)

jobrole_attrition = jobrole_attrition.sort_values()

print(jobrole_attrition.round(2))
```

```python
plt.figure(figsize=(10, 6))

jobrole_attrition.plot(kind="barh")

plt.title("Attrition Rate by Job Role")
plt.xlabel("Attrition Rate (%)")
plt.ylabel("Job Role")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(
    data["MonthlyIncome"],
    bins=20,
    kde=True
)

plt.title("Monthly Income Distribution")
plt.xlabel("Monthly Income")
plt.ylabel("Number of Employees")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    x="Attrition",
    y="Age",
    data=data
)

plt.title("Age Distribution by Attrition")
plt.xlabel("Attrition")
plt.ylabel("Age")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    x="Attrition",
    y="MonthlyIncome",
    data=data
)

plt.title("Monthly Income by Attrition")
plt.xlabel("Attrition")
plt.ylabel("Monthly Income")
plt.show()
```
